# MOE





MOE是拆开成多个，通过gate选择前topk个，然后softmax得到每个专家的权重，选中的专家输出乘以权重相加得到最终的输出。MOE是对每个token进行的，不是序列



显存占用不变且会有负载不均衡的现象产生。计算的更快了









deepseek的做法是让专家更加的专精，让专家总数目变得更多，而且自身的参数量进一步的减小。同时弄一个常驻的专家，用作所有专家的通用能力，在每次都需要激活。








softmax以后选择topk再softmax，那么第一次softmax实际上没有那么重要，所以V3改成了sigmoid



某个专家负载过高就减小偏置值bi否则就反之，这个偏置值只和专家路由有作用，在最终专家的输出相乘时候还是采用原来的权重值，不会加偏置值

V3改成了序列负载均衡损失函数，不再使用batch负载均衡（不然很可能是根据序列划分的专家，同种序列的问题太多，专家还是会负载）

# MTP


deepseekV3之前的MTP架构，用L-1的数据划分不同的transformer block，输出各自后续的内容，其中t2的损失权重大，t3小一点，以此类推。





通过预测和验证两部运算就出来四个token，原来需要四步，但是增加了GPU的运算量，因为第二部需要同时对三个序列进行预测。GPU擅长并行运算，主要时间开销是显存访问，所以三个序列生成和一个序列生成时间差不多



还可以将预测和验证放在一起，依次循环

但是MTP只在预测生成单个序列有效，也就是整个集群为一个人服务的时候，可以加速推理。因为现在vLLM用动态batch，会自动将多用户的不同请求组合成batch来预测，这时候很可能验证不通过，所以训练的时候用MTP来提升模型的性能，推理部署时一般就丢弃掉其他几个预测，变回单头大模型。





deepseek的MTP架构进行了更改，因为根据t1直接预测后面好几步的token有点太难了，所以把原来的信息以及t2的信息也整合在了一起，依次循环。注意这里是L层的信息了不再是L-1层。
