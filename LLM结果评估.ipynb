# PPL/perplexity


困惑度perplexity(PPL).每个token概率的几何平均数的倒数，越大困惑度越高 模型能力越差



这样计算更加稳定，因为seqlen其实很长，太多小数相乘就太小了 容易溢出，这样log再exp实际上不容易出错。

困惑度只能反应模型输出是否类似于测试文本，比如测试文本就是小学生作文，但是模型输出是顶尖作家，这样困惑度也会很高。但是不代表模型很弱
