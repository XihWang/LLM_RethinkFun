# mcp model context protocal

大模型只会问答，不会使用外部工具，mcp使得它获得了这个能力

MCP server 只是一个程序。不同的tools实际上是一个function，server本质上也是程序。可以认为server是类，tools是类函数。






其中mcp host就是cline



输入是cline的发送给server，输出是server返回给cline




MCP就是红框的内容

MCP是函数的注册和使用，和模型没有关系，没有提到和模型的交互方式。

MCP是函数的发现和调用协议！！！

模型上下文协议，上下文就是环境，环境里有很多函数来调用，来得到各种信息。

# function calling



蓝色是function call

注意更严格来说是红色部分是function call，但是我们一般都是用api来调用的，接触不到模型，所以我们认为蓝色部分才是func call。

这里openai服务器 实际上就是mcp host，网络搜索函数就是mcp server



两者不是互相替代的，而是相辅相成的

func call是模型与外部工具交互的能力，但是模型是不能调用工具（函数）的，所以需要一个mcp host来帮他调用
