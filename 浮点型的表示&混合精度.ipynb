低精度的好处：计算速度提升、显存占用变小、数据传输带宽占用小




低精度的问题：梯度一般比参数小，太小的梯度可能会变成0.大数吃小数、表示范围问题

混合精度下大部分计算都是在float16下进行的，但是我们还是会保存一份FP32的参数。防止梯度太小变成0，防止参数相对太大梯度被吃掉



参数先得降成FP16进行前向传播和后向传播的运算，然后再变成FP32传入优化器进行参数更新




梯度小 所以16位的范围有时候不够用 可以乘以loss进行偏移（因为loss变大N，梯度也会变大N）


一般求和 loss得是FP32的

卷积 transpose 矩阵乘法等是FP16的


乘法的时候16  累加的时候32  完成以后再把结果转化为16
