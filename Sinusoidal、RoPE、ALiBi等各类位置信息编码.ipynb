# sinusoidal


pos是seq的序号，比如Lpos是0，张pos是3。

对于4列hidden_sim，第0和第2列用sin，第1和第3列用cos

前期位置的变化率大，后期变化率就小了。一开始用于翻译，只需要上下文，所以后期GPT就有点不行了

# rope





加seqlen只要直接乘以对应的新的旋转因子就可以了，所以他外推性很好





横轴相对距离越大，那么相对上界（Q和K的内积，也就是注意力程度越小），体现了相对距离远的关注度减少

绝对位置编码针对qkv（因为直接加到X中的 所以V中在linear以后也有的）， 相对位置编码只需要qk就可



rope的周期比sinusoidal大很多，而且sinusoidal的加性嵌入会干扰。
